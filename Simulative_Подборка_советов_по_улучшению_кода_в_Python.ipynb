Ряд операций в коммерческом коде при обычном написании могут выглядеть достаточно громоздко и плохо читаемо. Это можно исправить за счет грамотного использования скобок, чему и будет посвящена первая часть нашей подборки.


# 1. Распаковка списка с помощью кортежа и скобок

Допустим, нам необходимо распаковать кортеж

```python
meats = ["chichken", "fish"]
```

```python
a, b = meats

print(a)
print(b)
```

Использование коротких названий переменных не приветствуется. Вместо этого используются более длинные, но более информативные названия. Обработать их можно с помощью кортежа и скобок

```python
(
    chicken_meat,
    fish_meat
) = meats

print(chicken_meat)
print(fish_meat)
```

Таким образом, мы можем использовать скобки для помощи в распаковке кортежей

Обратите внимание, что благодаря этому наша распаковка кортежей может вместить гораздо более длинные и более описательные имена переменных.

# 2. Логические условия со скобками

Допустим, у нас есть оператор if с 3 условиями, объединенными с помощью ключевого слова and.

```python
condition1, condition2, condition3 = True, True, True

if condition1 and condition2 and condition3:
    print("All conditions are True")
```

В рабочем коде условия становятся длиннее, и может быть больше условий. Поэтому один из способов решения этой проблемы - рефакторинг этого гигантского условия в отдельную функцию.
Или, если мы решим, что нет необходимости писать новую функцию только для этого, мы можем записать наши условные выражения, используя скобки.

```python
if (
    condition1
    and condition2
    and condition3
    ):
    print("All conditions are True")
```

Таким образом, мы не вынуждены писать новую функцию или переменную только для этого одного условного выражения, и при этом мы можем сохранить его аккуратным и читаемым.

# 3. Комбинирование строк с помощью скобок

Строки производственного уровня обычно не помещаются в одну строку из-за своей многословности. Поэтому мы объединяем их с помощью скобок.

```python
RETRY_INTERVAL = 10
var1 = "var1"
var2 = "var2"

error_log_message = (
    "ERROR. Failed to perform computaion"
    f"on {var1} and {var2}."
    f"Retrying in {RETRY_INTERVAL} seconds."
)

error_log_message
```

Примечание — внутри скобок строковые литералы, заключенные в кавычки, автоматически складываются, и нам не нужно использовать оператор + для этого.

# 4. Многострочное связывание методов с помощью скобок

В промышленном коде названия методов обычно намного длиннее, и мы часто связываем вместе больше методов.

И снова, мы используем скобки, чтобы разместить весь этот код на нескольких строках, вместо того чтобы сокращать названия методов или переменных.

```python
import pandas as pd

df = pd.DataFrame({"group_id": [1, 1, 2, 2], "name": ["Anton", "Elmira", "Sergey", "Elena"]})
df
```

```python
df.groupby(by=["group_id"]).agg({"name": "nunique"}).reset_index().sort_values(by="name", ascending=False)
```

Обратите внимание, что если мы выполняем цепочку методов внутри скобок, нам не требуется использовать символ \ для явного разрыва строки.

```python
output = (
    df
    .groupby(by=["group_id"])
    .agg({"name": "nunique"})
    .reset_index()
    .sort_values(by="name", ascending=False)
)
output
```
